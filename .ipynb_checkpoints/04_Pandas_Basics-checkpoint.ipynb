# 1. What is Pandas?

Pandas is a Python library used for **data manipulation and data analysis**.

It provides powerful data structures for working with structured and tabular data.

The two main data structures in Pandas are:

- Series
- DataFrame

Pandas is built on top of NumPy and is commonly used in:

- Data Analysis
- Data Science
- Machine Learning
- Data Cleaning
- Data Visualization
- Data Processing

# 2. Why Pandas?

Pandas makes it easier to work with structured data.

It provides functionality for:

- Reading data from files
- Writing data to files
- Cleaning data
- Filtering data
- Sorting data
- Handling missing values
- Grouping data
- Analyzing data
- Working with rows and columns

### Example

Without Pandas, working with large tabular datasets using only Python lists and dictionaries can become difficult.

Pandas provides DataFrames that make tabular data easier to work with.

# 3. Import Pandas

The standard way to import Pandas is:

`import pandas as pd`

The alias `pd` is commonly used throughout Python data analysis code.

```python
import pandas as pd

print(pd.__version__)
```

# 4. Series

A Pandas **Series** is a one-dimensional labeled data structure.

It can contain:

- Numbers
- Strings
- Boolean values
- Other Python objects

A Series contains:

- Values
- Index

```python
import pandas as pd

numbers = pd.Series([10, 20, 30, 40, 50])

print(numbers)
```

## Series with Custom Index

We can provide our own index labels when creating a Series.

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 75],
    index=["Alice", "Bob", "Charlie"]
)

print(marks)
```

# 5. Accessing Series Values

We can access Series values using their index.

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 75],
    index=["Alice", "Bob", "Charlie"]
)

print(marks["Alice"])
print(marks["Bob"])
```

# 6. Series Attributes

Some important Series attributes are:

- `values` → Returns the values
- `index` → Returns the index
- `dtype` → Returns the data type
- `size` → Returns the number of elements
- `shape` → Returns the shape

```python
import pandas as pd

marks = pd.Series([85, 90, 75, 80])

print("Values:", marks.values)
print("Index:", marks.index)
print("Data Type:", marks.dtype)
print("Size:", marks.size)
print("Shape:", marks.shape)
```

# 7. DataFrame

A **DataFrame** is a two-dimensional labeled data structure.

It is similar to a table in a database or a spreadsheet.

A DataFrame contains:

- Rows
- Columns
- Index
- Values

Example:

| Name | Age | Marks |
|---|---:|---:|
| Alice | 20 | 85 |
| Bob | 21 | 90 |
| Charlie | 19 | 75 |

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

print(df)
```

# 8. Creating DataFrames

There are several ways to create a DataFrame.

Common methods include:

- Dictionary
- List of Lists
- List of Dictionaries
- NumPy Array
- Existing files such as CSV and Excel files

## 8.1 DataFrame from a Dictionary

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

print(df)
```

## 8.2 DataFrame from a List of Lists

```python
import pandas as pd

data = [
    ["Alice", 20, 85],
    ["Bob", 21, 90],
    ["Charlie", 19, 75]
]

df = pd.DataFrame(
    data,
    columns=["Name", "Age", "Marks"]
)

print(df)
```

## 8.3 DataFrame from a List of Dictionaries

```python
import pandas as pd

data = [
    {"Name": "Alice", "Age": 20, "Marks": 85},
    {"Name": "Bob", "Age": 21, "Marks": 90},
    {"Name": "Charlie", "Age": 19, "Marks": 75}
]

df = pd.DataFrame(data)

print(df)
```

# 9. Data Types

Each column in a DataFrame has a data type.

Common Pandas data types include:

- `int64` → Integer values
- `float64` → Decimal values
- `object` → Usually text/string data
- `bool` → True/False values
- `datetime64` → Date and time values

We can use `dtypes` to check the data types of all columns.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85.5, 90.0, 75.5],
    "Passed": [True, True, True]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

## Checking the Data Type of a Single Column

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19]
}

df = pd.DataFrame(data)

print("Name dtype:", df["Name"].dtype)
print("Age dtype:", df["Age"].dtype)
```

# 10. Index

An **index** identifies each row in a DataFrame.

By default, Pandas assigns integer indexes starting from `0`.

Example:

| Index | Name | Age |
|---:|---|---:|
| 0 | Alice | 20 |
| 1 | Bob | 21 |
| 2 | Charlie | 19 |

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19]
}

df = pd.DataFrame(data)

print(df.index)
```

## Custom Index

We can specify custom index labels while creating a DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19]
}

df = pd.DataFrame(
    data,
    index=["Student1", "Student2", "Student3"]
)

print(df)
```

## Changing the Index

We can also assign a new index after creating the DataFrame.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19]
}

df = pd.DataFrame(data)

df.index = ["Student1", "Student2", "Student3"]

print(df)
```

# 11. Columns

Columns represent different variables or features in a DataFrame.

We can access all column names using the `columns` attribute.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

print(df.columns)
```

## Accessing a Single Column

A single column can be accessed using:

`df["column_name"]`

This returns a Pandas Series.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

print(df["Name"])
print(df["Marks"])
```

## Accessing Multiple Columns

We can access multiple columns by passing a list of column names.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

result = df[["Name", "Marks"]]

print(result)
```

# 12. Reading Data

Pandas can read data from many different file formats.

Common functions include:

- `pd.read_csv()` → Read CSV files
- `pd.read_excel()` → Read Excel files
- `pd.read_json()` → Read JSON files
- `pd.read_sql()` → Read data from SQL databases

The most commonly used function is `pd.read_csv()`.

# 13. Reading a CSV File

CSV stands for **Comma-Separated Values**.

A CSV file stores tabular data in rows and columns.

We can read a CSV file using:

`pd.read_csv()`

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

# 16. Writing Data

Pandas can save DataFrames to different file formats.

Common functions include:

- `to_csv()` → Write CSV files
- `to_excel()` → Write Excel files
- `to_json()` → Write JSON files

# 17. Writing to a CSV File

The `to_csv()` method is used to save a DataFrame as a CSV file.

### Syntax

`df.to_csv("filename.csv")`

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

df.to_csv("students1.csv", index=False)

print("CSV file created successfully.")
```
