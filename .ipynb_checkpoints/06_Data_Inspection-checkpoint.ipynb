# 1. Introduction to Data Inspection

Data inspection means checking and understanding a dataset before performing analysis.

It helps us understand:

- What the data looks like
- Number of rows and columns
- Column names
- Data types
- Missing values
- Statistical information
- Unique values
- Memory usage
- Distribution of categorical values

Pandas provides several functions and attributes for inspecting DataFrames.

# 2. Import Pandas

First, import the Pandas library.

```python
import pandas as pd
```

# 3. Create a Sample DataFrame

We will use the following DataFrame throughout this notebook.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma", "Frank", "Grace", "Henry"],
    "Age": [22, 25, 21, 28, 24, 30, 26, 23],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "IT", "Finance", "IT"
    ],
    "Salary": [50000, 45000, 55000, 60000, 48000, 70000, 65000, 52000],
    "Experience": [1, 2, 1, 5, 2, 7, 4, 2]
}

df = pd.DataFrame(data)

print(df)
```

# 4. head()

The `head()` method displays the first few rows of a DataFrame.

By default, it displays the first 5 rows.

### Syntax

`df.head()`

### Why use it?

It is useful for quickly checking:

- Column names
- First few records
- Data format
- Whether the data was loaded correctly

```python
df.head()
```

## Display a Specific Number of Rows

We can provide a number inside `head()`.

For example:

`df.head(3)`

displays the first 3 rows.

```python
df.head(3)
```

# 5. tail()

The `tail()` method displays the last few rows of a DataFrame.

By default, it displays the last 5 rows.

### Syntax

`df.tail()`

### Why use it?

It is useful for checking:

- The end of the dataset
- Whether data was loaded completely
- The last few records

```python
df.tail()
```

## Display a Specific Number of Last Rows

We can specify the number of rows.

For example:

`df.tail(3)`

displays the last 3 rows.

```python
df.tail(3)
```

# 6. sample()

The `sample()` method randomly selects rows from a DataFrame.

### Syntax

`df.sample()`

By default, it returns one random row.

### Why use it?

It is useful when:

- The dataset is very large
- We want to randomly inspect records
- We want to check whether the data contains unexpected values

```python
df.sample()
```

## Selecting Multiple Random Rows

We can specify the number of rows to randomly select.

For example:

`df.sample(3)`

selects 3 random rows.

```python
df.sample(3)
```

## Random Sample with a Fixed Seed

We can use `random_state` to get the same random sample every time.

This is useful when we want reproducible results.

```python
df.sample(3, random_state=42)
```

# 7. info()

The `info()` method provides a summary of the DataFrame.

It displays:

- Number of rows
- Number of columns
- Column names
- Number of non-null values
- Data types
- Memory usage

### Syntax

`df.info()`

### Why use it?

It is one of the most useful methods for quickly understanding a dataset.

```python
df.info()
```

## Understanding info()

For example:

`RangeIndex: 8 entries`

means the DataFrame contains 8 rows.

`8 non-null`

means all 8 rows contain a value for that column.

`int64`

means the column contains integer values.

`object`

usually indicates text/string data in many Pandas DataFrames.

`memory usage`

shows approximately how much memory the DataFrame uses.

# 8. describe()

The `describe()` method provides statistical summary information for numerical columns.

It commonly displays:

- count
- mean
- standard deviation
- minimum
- 25th percentile
- 50th percentile
- 75th percentile
- maximum

### Syntax

`df.describe()`

```python
df.describe()
```

## Understanding describe()

### count

Number of non-missing values.

### mean

Average value.

### std

Standard deviation, which measures the spread of the values.

### min

Smallest value.

### 25%

First quartile (Q1).

### 50%

Median (Q2).

### 75%

Third quartile (Q3).

### max

Largest value.

## Describe a Specific Column

We can also calculate descriptive statistics for a single column.

```python
df["Salary"].describe()
```

## Include Categorical Columns

By default, `describe()` focuses on numerical columns.

We can use:

`include="object"`

to describe object/string columns.

```python
df.describe(include="object")
```

## Describe All Columns

We can use:

`include="all"`

to generate descriptive information for both numerical and non-numerical columns.

```python
df.describe(include="all")
```

# 9. shape

The `shape` attribute returns the dimensions of a DataFrame.

It returns:

`(number_of_rows, number_of_columns)`

### Syntax

`df.shape`

### Example

If the result is:

`(8, 5)`

it means:

- 8 rows
- 5 columns

### Important

`shape` is an attribute, so we do not use parentheses.

```python
df.shape
```

## Number of Rows and Columns

We can unpack the shape into two variables.

```python
rows, columns = df.shape

print("Number of rows:", rows)
print("Number of columns:", columns)
```

# 10. columns

The `columns` attribute returns the names of all columns in a DataFrame.

### Syntax

`df.columns`

### Why use it?

It is useful for:

- Checking column names
- Selecting columns
- Renaming columns
- Understanding dataset structure

```python
df.columns
```

## Convert Column Names to a List

```python
list(df.columns)
```

# 11. dtypes

The `dtypes` attribute displays the data type of each column.

### Syntax

`df.dtypes`

Common data types include:

- `int64` → Integer
- `float64` → Decimal number
- `object` → Usually text/string
- `bool` → True/False
- `datetime64` → Date and time

### Why use it?

Data types are important because different operations work differently depending on the type of data.

```python
df.dtypes
```

## Check the Data Type of One Column

```python
print("Age:", df["Age"].dtype)
print("Salary:", df["Salary"].dtype)
print("Name:", df["Name"].dtype)
```

# 12. memory_usage()

The `memory_usage()` method shows how much memory is being used by each column.

### Syntax

`df.memory_usage()`

This is useful when working with large datasets.

### Why is memory usage important?

Large datasets can consume a lot of RAM.

Understanding memory usage helps us:

- Optimize datasets
- Reduce memory consumption
- Improve performance
- Work with larger datasets

```python
df.memory_usage()
```

## Include Object Memory Usage

For object/string columns, Pandas may not include the full size of the referenced objects by default.

We can use:

`deep=True`

to get a more detailed memory estimate.

```python
df.memory_usage(deep=True)
```

## Total Memory Usage

We can calculate the total memory used by the DataFrame.

```python
total_memory = df.memory_usage(deep=True).sum()

print("Total memory usage:", total_memory, "bytes")
```

# 13. nunique()

The `nunique()` method returns the number of unique values in each column.

### Syntax

`df.nunique()`

### Why use it?

It is useful for understanding:

- Number of unique categories
- Number of different values
- Categorical columns
- Potential duplicate values

```python
df.nunique()
```

## Unique Values in a Specific Column

```python
print("Unique departments:", df["Department"].nunique())
print("Unique ages:", df["Age"].nunique())
```

## `nunique()` with `dropna`

By default, missing values are not counted as unique values.

We can use:

`dropna=False`

to include missing values in the count.

```python
df.nunique(dropna=False)
```

# 14. value_counts()

The `value_counts()` method counts how frequently each unique value appears in a Series.

### Syntax

`df["column"].value_counts()`

### Why use it?

It is especially useful for categorical data.

For example, we can find:

- Number of employees in each department
- Number of students in each grade
- Number of customers in each city
- Number of products in each category

```python
df["Department"].value_counts()
```

## Example

If the Department column contains:

IT
IT
HR
Finance
IT

Then:

`value_counts()`

will show the number of occurrences of each department.

## value_counts() for Another Column

```python
df["Age"].value_counts()
```

## Sort by Index

By default, `value_counts()` sorts by frequency.

We can sort by the values themselves using:

`sort_index()`

```python
df["Department"].value_counts().sort_index()
```

## Convert Counts to Percentages

We can use:

`normalize=True`

to get the proportion of each value instead of the count.

```python
df["Department"].value_counts(normalize=True)
```

## Convert Counts to Percentages

We can multiply the normalized values by 100.

```python
department_percentage = (
    df["Department"]
    .value_counts(normalize=True) * 100
)

print(department_percentage)
```
