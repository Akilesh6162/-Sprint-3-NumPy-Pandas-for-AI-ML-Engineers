# Notebook 7 – Selecting & Filtering Data

This notebook covers different techniques used to select, filter, and sort data in Pandas.

## Topics Covered

- `loc`
- `iloc`
- Boolean Indexing
- Filtering Rows
- `query()`
- `isin()`
- `between()`
- Sorting Data

## 1. Import Pandas

Pandas is a Python library used for data manipulation and analysis.

````python
import pandas as pd
````

## Creating a Sample DataFrame

We will use the following DataFrame to understand data selection and filtering.

````python
data = {
    "Name": ["Arun", "Bala", "Charan", "Divya", "Elango", "Fathima"],
    "Age": [22, 25, 28, 21, 30, 26],
    "City": ["Chennai", "Bangalore", "Chennai", "Hyderabad", "Bangalore", "Chennai"],
    "Salary": [30000, 45000, 50000, 28000, 60000, 42000],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"]
}

df = pd.DataFrame(data)

df
````

## 2. loc

`loc` is used to select rows and columns using **labels**.

### Syntax

```python
df.loc[row_label, column_label]

```
It can be used to:

- Select specific rows
- Select specific columns
- Select a range of rows
- Filter rows using conditions

````python
# Select the first row

df.loc[0]
````

````python
# Select rows from index 0 to 2

df.loc[0:2]
````

````python
# Select a specific column

df.loc[:, "Name"]
````

````python
# Select multiple columns

df.loc[:, ["Name", "Salary"]]
````

````python
# Select specific rows and columns

df.loc[0:2, ["Name", "Salary"]]
````

````python
# Select rows where salary is greater than 40000

df.loc[df["Salary"] > 40000]
````

## 3. iloc

`iloc` is used to select rows and columns using **integer positions**.

### Syntax

```python
df.iloc[row_position, column_position]

````python

# Select the first row

df.iloc[0]
````

````python
# Select the first three rows

df.iloc[0:3]
````

````python
# Select the first column

df.iloc[:, 0]
````

````python
# Select the first two columns

df.iloc[:, 0:2]
````

````python
# Select rows 0 to 2 and columns 0 to 2

df.iloc[0:3, 0:3]
````

````python
# Select specific rows and columns

df.iloc[[0, 2, 4], [0, 3]]
````

## 4. Boolean Indexing

Boolean indexing is used to filter data based on a condition.

The condition returns `True` or `False` for each row.

### Example

```python
df["Age"] > 25

````python
# Check which employees are older than 25

df["Age"] > 25
````

````python
# Select employees older than 25

df[df["Age"] > 25]
````

````python
# Multiple conditions using AND (&)

df[(df["Age"] > 25) & (df["Salary"] > 40000)]
````

````python
# Multiple conditions using OR (|)

df[(df["City"] == "Chennai") | (df["City"] == "Bangalore")]
````

## 5. Filtering Rows

Filtering is used to select only the rows that satisfy a particular condition.

Common filtering operators:

- `>`
- `<`
- `>=`
- `<=`
- `==`
- `!=`
- `&` – AND
- `|` – OR

````python
# Filter employees from Chennai

df[df["City"] == "Chennai"]
````

````python
# Filter employees from IT department

df[df["Department"] == "IT"]
````

````python
# Filter employees with salary less than or equal to 40000

df[df["Salary"] <= 40000]
````

````python
# Filter employees between age 25 and 30

df[(df["Age"] >= 25) & (df["Age"] <= 30)]
````

## 6. query()

`query()` is used to filter rows using a string expression.

It provides a cleaner way to write some filtering conditions.

### Syntax

```python
df.query("condition")

````python
# Employees older than 25

df.query("Age > 25")
````

````python
# Employees from Chennai

df.query("City == 'Chennai'")
````

````python
# Multiple conditions

df.query("Age > 25 and Salary > 40000")
````

## 7. isin()

`isin()` is used to check whether values belong to a specified list of values.

It is useful when we want to filter rows based on multiple possible values.

### Syntax

```python
df["column"].isin([value1, value2, value3])

````python
# Employees from Chennai or Bangalore

df[df["City"].isin(["Chennai", "Bangalore"])]
````

````python
# Employees from IT or HR departments

df[df["Department"].isin(["IT", "HR"])]
````

````python
# Check whether each city is Chennai or Hyderabad

df["City"].isin(["Chennai", "Hyderabad"])
````

## 8. between()

`between()` is used to check whether values fall within a specified range.

### Syntax

```python
df["column"].between(lower, upper)

````python
# Employees whose age is between 25 and 30

df[df["Age"].between(25, 30)]
````

````python
# Employees whose salary is between 30000 and 50000

df[df["Salary"].between(30000, 50000)]
````

````python
# Age between 25 and 30, excluding the boundaries

df[df["Age"].between(25, 30, inclusive="neither")]
````

## 9. Sorting Data

Sorting is used to arrange DataFrame rows in ascending or descending order.

The main method used is:

```python
sort_values()

````python
df.sort_values("Name")
````

````python
# Sort by salary in ascending order

df.sort_values("Salary", ascending=True)
````

````python
# Sort by salary in descending order

df.sort_values("Salary", ascending=False)
````

````python
# Sort by multiple columns

df.sort_values(["Department", "Salary"])
````

````python
# Sort department ascending and salary descending

df.sort_values(
    ["Department", "Salary"],
    ascending=[True, False]
)
````

## 10. Sorting by Index

`sort_index()` is used to sort the DataFrame based on its index.

````python
# Sort by index

df.sort_index()
````

````python
# Sort index in descending order

df.sort_index(ascending=False)
````

## 12. Difference Between loc and iloc

| Feature | loc | iloc |
|---|---|---|
| Selection type | Label-based | Position-based |
| Uses | Row/column labels | Integer positions |
| Example | `df.loc[0]` | `df.iloc[0]` |
| Column selection | `df.loc[:, "Name"]` | `df.iloc[:, 0]` |
| Filtering | Supports conditions | Mainly position-based selection |

### Simple Rule

- Use `loc` when you know the **label/name**.
- Use `iloc` when you know the **position/index number**.

# Summary

In this notebook, we learned:

- `loc` – Select data using labels.
- `iloc` – Select data using integer positions.
- Boolean Indexing – Filter data using conditions.
- Filtering Rows – Select rows that satisfy conditions.
- `query()` – Filter rows using expressions.
- `isin()` – Filter rows based on multiple values.
- `between()` – Filter values within a range.
- `sort_values()` – Sort data by one or more columns.
- `sort_index()` – Sort data based on the index.

These techniques are commonly used in Pandas for selecting, filtering, and organizing data during data analysis.
