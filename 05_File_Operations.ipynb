# 3. CSV Files

CSV stands for **Comma-Separated Values**.

CSV files store data in rows and columns.

Example:

Name,Age,Marks
Alice,20,85
Bob,21,90
Charlie,19,75

CSV files are commonly used because they are:

- Simple

# 4. Reading CSV Files

Pandas provides `pd.read_csv()` to read CSV files.

### Syntax

`pd.read_csv("filename.csv")`

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

# 5. Creating a CSV File

We can create a DataFrame and save it as a CSV file using `to_csv()`.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

# 6. Reading CSV with Different Options

`read_csv()` provides many useful parameters.

Some commonly used parameters are:

- `sep` → Separator
- `header` → Header row
- `usecols` → Select specific columns
- `nrows` → Read a specific number of rows
- `index_col` → Set a column as index

```python
import pandas as pd

df = pd.read_csv(
    "students.csv",
    usecols=["Name", "Marks"]
)

print(df)
```

# 11. Writing Multiple Sheets to Excel

We can save multiple DataFrames into different sheets of the same Excel file.

`ExcelWriter` is used for this purpose.

```python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Marks": [85, 90, 75]
})

employees = pd.DataFrame({
    "Name": ["John", "David", "Emma"],
    "Salary": [50000, 60000, 55000]
})

with pd.ExcelWriter("data.xlsx") as writer:
    students.to_excel(writer, sheet_name="Students", index=False)
    employees.to_excel(writer, sheet_name="Employees", index=False)

print("Excel file with multiple sheets created.")
```

# 14. Creating a JSON File

We can use `to_json()` to save a DataFrame as JSON.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

df.to_json(
    "students.json",
    orient="records",
    indent=4
)

print("JSON file created successfully.")
```

# 15. JSON Orient

The `orient` parameter controls how the DataFrame is represented in JSON.

A commonly used option is:

`orient="records"`

This represents each DataFrame row as a JSON object.

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob"],
    "Marks": [85, 90]
}

df = pd.DataFrame(data)

print(
    df.to_json(
        orient="records",
        indent=4
    )
)
```

# 19. Pickle Files

Pickle is a Python-specific serialization format.

It can store Python objects so that they can be loaded later.

Pandas provides:

- `read_pickle()`
- `to_pickle()`

Pickle is useful when we want to save and restore Python/Pandas objects while preserving their Python-specific structure and data types.

### Important

Only load Pickle files from trusted sources.

Pickle data can execute arbitrary code when loaded.

# 20. Writing a Pickle File

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
}

df = pd.DataFrame(data)

df.to_pickle("students.pkl")

print("Pickle file created successfully.")
```

# 22. SQL Tables

SQL databases store data in tables.

Pandas can read data from SQL databases into a DataFrame.

Common SQL databases include:

# 23. SQLite Example

SQLite is a lightweight database that stores the entire database in a file.

Python provides the built-in `sqlite3` module for working with SQLite.

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("students.db")

data = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 75]
})

data.to_sql(
    "students",
    connection,
    if_exists="replace",
    index=False
)

print("Table created successfully.")
```

# 24. Reading a SQL Table

We can use `pd.read_sql()` to read a SQL query into a DataFrame.

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("students.db")

query = "SELECT * FROM students"

df = pd.read_sql(
    query,
    connection
)

print(df)

connection.close()
```

# 25. Writing DataFrame to SQL

The `to_sql()` method can save a DataFrame as a SQL table.

### Important Parameters

- `name` → Table name
- `con` → Database connection
- `if_exists` → What to do if the table already exists
- `index` → Whether to save the DataFrame index

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("employees.db")

employees = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 45000, 60000]
})

employees.to_sql(
    "employees",
    connection,
    if_exists="replace",
    index=False
)

print("Data saved to SQL table.")

connection.close()
```

# 28. Using `glob` to Find Multiple CSV Files

The `glob` module can find files matching a pattern.

```python
import glob

files = glob.glob("data/*.csv")

print(files)
```

# 29. Reading Multiple CSV Files

We can loop through the files, read each CSV file, and store the DataFrames in a list.

```python
import pandas as pd
import glob

files = glob.glob("data/*.csv")

dataframes = []

for file in files:
    df = pd.read_csv(file)
    dataframes.append(df)

print(f"Number of files: {len(dataframes)}")
```

# 34. Checking Whether a File Exists

We can use `.exists()` to check whether a file or directory exists.

```python
from pathlib import Path

file_path = Path("students.csv")

if file_path.exists():
    print("File exists.")
else:
    print("File does not exist.")
```

# 35. File Name and Extension

`pathlib` provides useful properties for extracting parts of a file path.

```python
from pathlib import Path

path = Path("data/students.csv")

print("Full path:", path)
print("File name:", path.name)
print("File stem:", path.stem)
print("Extension:", path.suffix)
print("Parent folder:", path.parent)
```

# 36. Creating Directories

`Path.mkdir()` can be used to create a directory.

```python
from pathlib import Path

folder = Path("data")

folder.mkdir(
    exist_ok=True
)

print("Directory created or already exists.")
```

# 37. Finding Files Using pathlib

`Path.glob()` can be used to find files matching a pattern.

```python
from pathlib import Path

data_folder = Path("data")

csv_files = list(data_folder.glob("*.csv"))

for file in csv_files:
    print(file)
```
