# 1. What is NumPy?

NumPy stands for **Numerical Python**.

It is a Python library used for numerical and scientific computing.

NumPy provides a powerful data structure called an **ndarray (N-dimensional array)** that allows us to store and process large amounts of numerical data efficiently.

NumPy is commonly used in:

- Data Science
- Machine Learning
- Artificial Intelligence
- Scientific Computing
- Data Analysis
- Statistical Computing

# 2. Why NumPy?

NumPy is preferred over normal Python lists for numerical operations because it is:

- Faster
- Memory efficient
- Easy to use for mathematical operations
- Suitable for handling large datasets
- Supports multi-dimensional arrays
- Provides many mathematical functions
- Supports vectorized operations

### Example

With a Python list:

```python
numbers = [1, 2, 3, 4, 5]

result = [x * 2 for x in numbers]

print(result)
```

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5])

result = numbers * 2

print(result)
```

# 3. NumPy Arrays (`ndarray`)

The main data structure provided by NumPy is called an **ndarray**.

`ndarray` means **N-dimensional array**.

An array can contain:

- One-dimensional data
- Two-dimensional data
- Three-dimensional data
- Higher-dimensional data

Example:

1D array:

`[10, 20, 30, 40]`

2D array:

`[[10, 20], [30, 40]]`

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr)
print(type(arr))
```

# 4. Creating Arrays

NumPy provides several ways to create arrays.

Common methods include:

- `np.array()`
- `np.zeros()`
- `np.ones()`
- `np.full()`
- `np.arange()`
- `np.linspace()`
- `np.eye()`

```python
import numpy as np

# Using np.array()
arr1 = np.array([1, 2, 3, 4])

# Array of zeros
arr2 = np.zeros(5)

# Array of ones
arr3 = np.ones(5)

# Array with a specific value
arr4 = np.full(5, 10)

# Array using arange()
arr5 = np.arange(1, 10)

# Evenly spaced values
arr6 = np.linspace(0, 10, 5)

# Identity matrix
arr7 = np.eye(3)

print("Array:", arr1)
print("Zeros:", arr2)
print("Ones:", arr3)
print("Full:", arr4)
print("Arange:", arr5)
print("Linspace:", arr6)
print("Identity Matrix:\n", arr7)
```

# 5. Dimensions

The number of dimensions of an array is called its **dimension**.

NumPy uses the `ndim` attribute to find the number of dimensions.

### Examples

- 1D → `[1, 2, 3]`
- 2D → `[[1, 2], [3, 4]]`
- 3D → Collection of 2D arrays

```python
import numpy as np

arr1 = np.array([1, 2, 3])

arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D:", arr1.ndim)
print("2D:", arr2.ndim)
print("3D:", arr3.ndim)
```

# 6. Shape

The **shape** of an array tells us the size of the array along each dimension.

We use the `shape` attribute.

For a 2D array:

- First value = number of rows
- Second value = number of columns

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr)
print("Shape:", arr.shape)
```

The output is:

`(2, 3)`

This means the array has:

- 2 rows
- 3 columns

# 7. Size

The `size` attribute returns the **total number of elements** in an array.

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Total elements:", arr.size)
```

# 8. Data Types

NumPy arrays have a specific data type for their elements.

The `dtype` attribute is used to check the data type.

Common NumPy data types include:

- `int32`
- `int64`
- `float32`
- `float64`
- `bool`
- `str`

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([1.5, 2.5, 3.5])
arr3 = np.array([True, False, True])

print("Integer:", arr1.dtype)
print("Float:", arr2.dtype)
print("Boolean:", arr3.dtype)
```

## Specifying Data Type

We can explicitly specify the data type using the `dtype` parameter.

```python
import numpy as np

arr = np.array([1, 2, 3, 4], dtype=float)

print(arr)
print(arr.dtype)
```

# 9. Indexing

Indexing is used to access individual elements from an array.

Python uses **zero-based indexing**.

That means:

- First element → index `0`
- Second element → index `1`
- Third element → index `2`

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])
print(arr[2])
print(arr[4])
```

## Negative Indexing

Negative indexing allows us to access elements from the end.

- Last element → `-1`
- Second-last element → `-2`

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Last:", arr[-1])
print("Second last:", arr[-2])
```

## Indexing in 2D Arrays

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print(arr[0, 0])
print(arr[0, 2])
print(arr[1, 1])
```

# 10. Slicing

Slicing is used to extract a portion of an array.

Syntax:

`array[start:stop:step]`

The `stop` index is not included.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])
print(arr[:3])
print(arr[2:])
print(arr[::2])
```

## Slicing 2D Arrays

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[0:2, 0:2])
```

# 11. Reshaping

Reshaping changes the shape of an array without changing its data.

We use the `reshape()` method.

The total number of elements must remain the same.

For example:

A `6`-element array can be reshaped into:

- `(2, 3)`
- `(3, 2)`
- `(1, 6)`
- `(6, 1)`

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
print("Shape:", new_arr.shape)
```

# 12. Flatten

`flatten()` converts a multi-dimensional array into a one-dimensional array.

It returns a **copy** of the original array.

```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

flat = arr.flatten()

print("Original:")
print(arr)

print("\nFlattened:")
print(flat)
```

# 13. Copy vs View

NumPy arrays can be copied in two ways:

### Copy

A copy creates a completely independent array.

Changes to the copy do not affect the original array.

### View

A view creates another way of looking at the same underlying data.

Changes to the view can affect the original array.

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

# Copy
copy_arr = arr.copy()

copy_arr[0] = 100

print("Original:", arr)
print("Copy:", copy_arr)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

# View
view_arr = arr.view()

view_arr[0] = 100

print("Original:", arr)
print("View:", view_arr)
```

# 14. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of different shapes when their dimensions are compatible.

It allows us to perform operations without manually changing the shape of the arrays.

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

result = arr + 5

print(result)
```

Here, the value `5` is automatically applied to every element:

`[10, 20, 30, 40] + 5`

becomes:

`[15, 25, 35, 45]`

```python
import numpy as np

arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

result = arr + np.array([10, 20, 30])

print(result)
```

# 15. Array Operations

NumPy supports mathematical operations directly on arrays.

Common operations include:

- Addition
- Subtraction
- Multiplication
- Division
- Power
- Modulus

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
print("Modulus:", a % b)
```

## Mathematical Functions

NumPy also provides functions for mathematical calculations.

```python
import numpy as np

arr = np.array([1, 4, 9, 16, 25])

print("Square root:", np.sqrt(arr))
print("Square:", np.square(arr))
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
```

# 16. Universal Functions (ufuncs)

Universal functions, also called **ufuncs**, are NumPy functions that operate element-by-element on arrays.

They are optimized for fast numerical operations.

Examples:

- `np.sqrt()`
- `np.exp()`
- `np.sin()`
- `np.cos()`
- `np.log()`
- `np.abs()`
- `np.add()`
- `np.subtract()`
- `np.multiply()`
- `np.divide()`

```python
import numpy as np

arr = np.array([1, 4, 9, 16])

print("Square root:", np.sqrt(arr))
print("Exponential:", np.exp(arr))
print("Absolute:", np.abs([-10, -5, 0, 5, 10]))
```

## Ufuncs for Arithmetic Operations

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

print("Add:", np.add(a, b))
print("Subtract:", np.subtract(a, b))
print("Multiply:", np.multiply(a, b))
print("Divide:", np.divide(a, b))
```

# 17. Complete Example

The following example combines several NumPy concepts.

```python
import numpy as np

# Create an array
data = np.array([10, 20, 30, 40, 50, 60])

# Display basic information
print("Array:", data)
print("Dimensions:", data.ndim)
print("Shape:", data.shape)
print("Size:", data.size)
print("Data Type:", data.dtype)

# Indexing
print("\nFirst element:", data[0])
print("Last element:", data[-1])

# Slicing
print("First three elements:", data[:3])

# Reshaping
matrix = data.reshape(2, 3)

print("\nReshaped Array:")
print(matrix)

# Mathematical operation
print("\nArray multiplied by 2:")
print(data * 2)

# Mean
print("\nMean:", np.mean(data))

# Maximum
print("Maximum:", np.max(data))

# Minimum
print("Minimum:", np.min(data))
```
