# 2. Random Numbers

Random numbers are numbers generated in an unpredictable or pseudo-random manner.

NumPy provides functions for generating random numbers.

`np.random.rand()` generates random floating-point numbers between `0` and `1`.

### Where it is used

Random numbers are commonly used in:

- Machine Learning
- Simulations
- Data Science
- Testing
- Games
- Statistical experiments

```python
import numpy as np

number = np.random.rand()

print("Random number:", number)
```

```python
import numpy as np

numbers = np.random.rand(5)

print("Random numbers:", numbers)
```

## Random Numbers in a 2D Array

We can provide multiple dimensions to `np.random.rand()`.

For example:

`np.random.rand(2, 3)`

creates a 2 × 3 array containing random values between 0 and 1.

```python
import numpy as np

matrix = np.random.rand(2, 3)

print(matrix)
```

# 3. Random Seed

A random seed is used to make random results **reproducible**.

Normally, every time we generate random numbers, we may get different values.

By setting the same seed, we can generate the same sequence of random numbers.

NumPy provides:

`np.random.seed()`

```python
import numpy as np

np.random.seed(42)

numbers = np.random.rand(5)

print(numbers)
```

If we run the previous code multiple times with the same seed `42`, we get the same random numbers.

### Why is Random Seed useful?

Random seeds are useful when:

- Reproducing experiments
- Testing code
- Training machine learning models
- Comparing results
- Debugging random processes

# 4. Random Integers

`np.random.randint()` generates random integer values.

### Syntax

`np.random.randint(low, high, size)`

- `low` → Starting value (included)
- `high` → Ending value (excluded)
- `size` → Number of values to generate

```python
import numpy as np

number = np.random.randint(1, 10)

print("Random integer:", number)
```

```python
import numpy as np

numbers = np.random.randint(1, 10, size=5)

print("Random integers:", numbers)
```

In this example:

`np.random.randint(1, 10, size=5)`

generates 5 random integers from `1` to `9`.

The value `10` is not included.

## Random Integer Matrix

```python
import numpy as np

matrix = np.random.randint(1, 100, size=(3, 4))

print(matrix)
```

# 5. Random Choice

`np.random.choice()` randomly selects values from an array or sequence.

### Syntax

`np.random.choice(array, size)`

It can be used to randomly select one or more values.

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

result = np.random.choice(numbers)

print("Randomly selected value:", result)
```

```python
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

result = np.random.choice(numbers, size=3)

print("Randomly selected values:", result)
```

## Random Choice from Strings

`np.random.choice()` can also be used with strings.

```python
import numpy as np

colors = ["Red", "Blue", "Green", "Yellow"]

choice = np.random.choice(colors)

print("Selected color:", choice)
```

## Random Choice with Probability

We can specify the probability of selecting each value using the `p` parameter.

The probabilities must add up to `1`.

```python
import numpy as np

colors = ["Red", "Blue", "Green"]

choice = np.random.choice(
    colors,
    size=10,
    p=[0.5, 0.3, 0.2]
)

print(choice)
```

Here:

- Red has a probability of `50%`
- Blue has a probability of `30%`
- Green has a probability of `20%`

The result is still random, so the exact number of times each value appears can vary.

# 6. Shuffle

`np.random.shuffle()` randomly rearranges the elements of an array.

It modifies the **original array**.

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5])

print("Before shuffle:", numbers)

np.random.shuffle(numbers)

print("After shuffle:", numbers)
```

### Important

`shuffle()` changes the original array.

Therefore, if you need to preserve the original array, make a copy first.

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5])

shuffled = numbers.copy()

np.random.shuffle(shuffled)

print("Original:", numbers)
print("Shuffled:", shuffled)
```

# 7. Permutation

`np.random.permutation()` returns a randomly permuted copy of an array.

Unlike `shuffle()`, it does **not modify the original array**.

```python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5])

permuted = np.random.permutation(numbers)

print("Original:", numbers)
print("Permutation:", permuted)
```

## Shuffle vs Permutation

| Function | Original Array Modified? | Returns Result? |
|---|---|---|
| `np.random.shuffle()` | Yes | No |
| `np.random.permutation()` | No | Yes |

### Example

`shuffle()` → changes the original array.

`permutation()` → creates a randomly reordered copy.

# 8. Random Distributions

A probability distribution describes how random values are distributed.

NumPy provides functions for generating random values from different probability distributions.

Common distributions include:

- Uniform Distribution
- Normal Distribution
- Binomial Distribution
- Poisson Distribution
- Exponential Distribution
- Bernoulli Distribution

# 8.1 Uniform Distribution

In a uniform distribution, values within a specified range have equal likelihood.

NumPy provides:

`np.random.uniform()`

### Syntax

`np.random.uniform(low, high, size)`

```python
import numpy as np

numbers = np.random.uniform(1, 10, size=5)

print(numbers)
```

### Example

Generate 10 random values between 0 and 100.

```python
import numpy as np

marks = np.random.uniform(0, 100, size=10)

print(marks)
```

# 8.2 Normal Distribution

The normal distribution is a common probability distribution with a bell-shaped curve.

It is described using:

- Mean
- Standard deviation

NumPy provides:

`np.random.normal()`

### Syntax

`np.random.normal(mean, standard_deviation, size)`

```python
import numpy as np

numbers = np.random.normal(
    loc=50,
    scale=10,
    size=10
)

print(numbers)
```

Here:

- `loc=50` → Mean
- `scale=10` → Standard deviation
- `size=10` → Number of values

### Where it is used

Normal distributions are commonly used to model measurements such as heights, test scores, measurement errors, and many naturally occurring variables.

# 8.3 Binomial Distribution

The binomial distribution models the number of successes in a fixed number of independent trials.

NumPy provides:

`np.random.binomial()`

### Parameters

- `n` → Number of trials
- `p` → Probability of success
- `size` → Number of experiments

```python
import numpy as np

results = np.random.binomial(
    n=10,
    p=0.5,
    size=10
)

print(results)
```

### Example

Suppose a coin is tossed 10 times.

- Number of trials = 10
- Probability of getting heads = 0.5

The binomial distribution can tell us how many heads occurred in each simulated experiment.

# 8.4 Poisson Distribution

The Poisson distribution models the number of events occurring within a fixed interval of time or space when events occur at an average rate.

NumPy provides:

`np.random.poisson()`

### Parameter

- `lam` → Average number of events
- `size` → Number of samples

```python
import numpy as np

events = np.random.poisson(
    lam=5,
    size=10
)

print(events)
```

### Example

A website receives an average of 5 customer requests per minute.

The Poisson distribution can be used to simulate the number of requests received during different minutes.

# 8.5 Exponential Distribution

The exponential distribution is commonly used to model the time between events in a Poisson process.

NumPy provides:

`np.random.exponential()`

### Parameter

- `scale` → Average waiting time
- `size` → Number of samples

```python
import numpy as np

waiting_times = np.random.exponential(
    scale=5,
    size=10
)

print(waiting_times)
```

### Example

If customers arrive at an average interval of 5 minutes, an exponential distribution can be used to simulate the waiting time until the next customer arrives.

# 8.6 Bernoulli Distribution

The Bernoulli distribution represents a single experiment with two possible outcomes:

- Success → 1
- Failure → 0

For example:

- Coin toss
- Pass or fail
- Yes or no
- Click or no click

NumPy can simulate Bernoulli trials using the binomial distribution with `n=1`.

```python
import numpy as np

results = np.random.binomial(
    n=1,
    p=0.5,
    size=10
)

print(results)
```

# 9. Random Distribution Comparison

| Distribution | Main Use |
|---|---|

| Normal | Bell-shaped continuous data |
| Binomial | Number of successes in fixed trials |
| Poisson | Number of events in a fixed interval |
| Exponential | Time between events |
| Bernoulli | Single success/failure experiment |
